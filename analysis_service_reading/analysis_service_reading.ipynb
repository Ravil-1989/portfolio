## Выполнение соответствующих запросов

### Посчитаем количество книг, которые вышли после 1 января 2000 года

```python
query = ''' SELECT COUNT(book_id)
            FROM books
            WHERE publication_date > '2000-01-01' '''
con=engine.connect()
pd.io.sql.read_sql(sql=text(query), con = con)
```

### Для каждой книги посчитаем количество обзоров и среднюю оценку

```python
query = '''SELECT b.title,
                  COUNT(DISTINCT rev.review_id) as count_review,
                  AVG(rat.rating) as avg_rating
           FROM books as b
           LEFT OUTER JOIN reviews as rev ON b.book_id = rev.book_id
           LEFT OUTER JOIN ratings as rat ON b.book_id = rat.book_id
           GROUP BY b.book_id
           ORDER BY count_review DESC'''
con=engine.connect()
pd.io.sql.read_sql(sql=text(query), con = con)
```

### Определим издательство, которое выпустило наибольшее число книг толще 50 страниц

```python
query = '''SELECT p.publisher
           FROM publishers as p
           LEFT OUTER JOIN books as b ON p.publisher_id = b.publisher_id
           WHERE b.num_pages > 50
           GROUP BY p.publisher
           ORDER BY COUNT(b.book_id) DESC
           LIMIT 1'''
con=engine.connect()
pd.io.sql.read_sql(sql=text(query), con = con)
```

### Определим автора с самой высокой средней оценкой книг, учитывая только книги с 50 и более оценками

```python
query = '''SELECT author,
                  AVG(avg_rating) as average_rating
           FROM(SELECT a.author,
                       b.author_id,
                       AVG(rat.rating) as avg_rating
                FROM books as b
                LEFT OUTER JOIN ratings as rat ON b.book_id = rat.book_id
                LEFT OUTER JOIN authors as a ON a.author_id = b.author_id
                GROUP BY b.author_id, b.book_id, a.author
                HAVING COUNT(rat.rating_id) >= 50
                ORDER BY avg_rating DESC) as avg_data
           GROUP BY author
           ORDER BY average_rating DESC
           LIMIT 1'''
con=engine.connect()
pd.io.sql.read_sql(sql=text(query), con = con)
```

### Посчитаем среднее количество обзоров от пользователей, которые поставили больше 48 оценок

```python
query = '''SELECT AVG(reviews_count) as avg_reviews
           FROM (SELECT DISTINCT rev.username,
                        COUNT(DISTINCT rev.review_id) AS reviews_count,
                        COUNT(DISTINCT rat.rating_id) AS ratings_count
                 FROM reviews as rev
                 LEFT OUTER JOIN ratings as rat ON rev.username = rat.username
                 GROUP BY rev.username
                 HAVING COUNT(DISTINCT rat.rating_id) > 48
                 ORDER BY ratings_count) as acive_users'''
con=engine.connect()
pd.io.sql.read_sql(sql=text(query), con = con)
```
